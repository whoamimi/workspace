## Method 3

Uses the same HTML output from Method 2.

```python
import pandas as pd
from io import StringIO

df = pd.read_html(StringIO(html))[0]

away = df.iloc[::2].reset_index(drop=True)
home = df.iloc[1::2].reset_index(drop=True)


# Split "(-1.5) +146" or "(O: 7) -113" into line and odds
def line_odds(s):
    return s.str.extract(r"\(?[OU:\s]*([+-]?\d+\.?\d*)\)?\s*([+-]\d+)").astype(float)


games = pd.DataFrame(
    {
        "date": pd.to_datetime(
            away["DATE"]
            .str.replace(r"^\w+,\s*", "", regex=True)  # drop weekday
            .str.replace(r"\s+", " ", regex=True)  # collapse double spaces
            .str.strip()
            + " 2026",
            format="%b. %d, %I:%M %p %Y",
        ),
        "away_team": away["MATCHUP"].str.removesuffix(" at").str.strip(),
        "home_team": home["DATE"],
        "away_ml": away["MONEY"].astype(int),
        "home_ml": home["SPREAD"].astype(int),
    }
)

games[["away_spread", "away_spread_odds"]] = line_odds(away["SPREAD"]).values
games[["home_spread", "home_spread_odds"]] = line_odds(home["MATCHUP"]).values
games[["total", "over_odds"]] = line_odds(away["TOTAL"]).values
games["under_odds"] = line_odds(home["MONEY"])[1].values
```
